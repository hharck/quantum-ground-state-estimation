```python
from QSE.Backends_magne import QiskitBackend
from QSE.QSE import QSE
import numpy as np

backend = QiskitBackend(
    backend_type="magne",
    shots=2000,
    noise=None,
    seed=1234,
)

qse = QSE(
    backend=backend,
    molecule="H2",
    num_trotter_steps=2,
    krylov_dim=2,
)

qse.run()

result = qse.job.result()[0]
print("Estimator values shape:", result.data.evs.shape)
print(result.data.evs)

energies = qse.energy_estimates()
print("QSE energies:", energies)
```
